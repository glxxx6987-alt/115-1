```python
a={"蘋果":50 ,"香蕉":30 ,"牛奶":60 ,"麵包":40 }
print("===== 超市商品 =====")
for i,j in a.items():
  print(f"{i}:{j}元")
total=0
while True:
  n=input("請輸入商品名稱（輸入 / 結束）：")
  if n=="/":
    break
  if n in a:
        total+=a[n]
  else:
      print("輸入不存在")
print(f"總金額:{total}元")
```
