```python
import math
r=float(input("請輸入角度:"))
d=r*(math.pi/180)
s=math.sin(d)
print(f"sin({r})={s:.2f}")
```

```python
import random

won = False
for d in range(1,4):
    num=int(input(f"今天第{d}天,請輸入1-50數字:"))
    lucky=random.randint(1,50)
    print(f"今天的幸運數字是{lucky}")
    if num == lucky:
        print("恭喜你中獎了！")
        won=True
        break
    else:
        if d<3:
            print("你沒中獎，明天再來\n")

if not won:
    print("很遺憾3天都沒中獎，下次再試試吧！")
```
