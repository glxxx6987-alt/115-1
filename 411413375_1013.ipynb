```python
def calculate_bmi(h,w):
    bmi = w/(h/100) ** 2

    print(f"{bmi:.2f}")
    return round(bmi, 2)

calculate_bmi(170, 60)
calculate_bmi(165, 55)
calculate_bmi(180, 80)
```
