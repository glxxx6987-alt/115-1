```python
listA=[]
while True:
  item=input("請輸入A倉文具,輸入'/'結束:")
  if item=="/":
    break
  listA.append(item)

listB=[]
while True:
  item=input("請輸入B倉文具,輸入'/'結束:")
  if item=="/":
    break
  listA.append(item)
listC=listA+listB
listC.append("膠水")
listC="/".join(listC)
print(f"二倉合併清單:{listC}")
```
